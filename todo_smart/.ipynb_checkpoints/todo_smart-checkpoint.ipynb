```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame(columns=['todo', 'priority', 'due_date'])
```

```python
import os
import json
import requests
from datetime import datetime

text = input("Describe the todo: ")

token = os.environ["GITHUB_TOKEN"]
today = datetime.now().isoformat()

prompt = f"""
Convert this natural-language todo into JSON with exactly these fields:
title (string), priority (one of low, medium, high), due_date (ISO 8601 datetime or null).

Current date and time: {today}
Todo: {text}
Return JSON only.
"""

response = requests.post(
    "https://models.github.ai/inference/chat/completions",
    headers={
        "Authorization": f"Bearer {token}",
        "Content-Type": "application/json",
    },
    json={
        "model": "openai/gpt-4o-mini",
        "messages": [{"role": "user", "content": prompt}],
        "temperature": 0,
    },
)
response.raise_for_status()

todo = json.loads(response.json()["choices"][0]["message"]["content"])
meta = [todo]

df1.loc[len(df1)] = [
    todo["title"],
    todo["priority"],
    todo["due_date"],
]
```
